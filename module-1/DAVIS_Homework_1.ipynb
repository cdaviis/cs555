```python
# Setup

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm

%matplotlib inline
```

```python
# Question 1

hospital = pd.read_csv("hospital_stays.csv")

days = hospital["hospital_stay_days"]

print("n =", len(days))
print("total days =", days.sum())
print("min =", days.min())
print("max =", days.max())
```

```python
# Question 2

plt.hist(days, bins=np.arange(1.5, 16.5, 1), edgecolor="black")
plt.xticks(range(2, 16))
plt.xlabel("Hospital Stay Duration in Days")
plt.ylabel("Number of Patients")
plt.title("Hospital Stay Durations")
plt.tight_layout()
plt.savefig("hospital_stays_hist.png")
plt.show()

q1 = days.quantile(0.25)
q3 = days.quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outliers = days[(days < lower) | (days > upper)]

print("IQR fences:", lower, "to", upper)
print("Values outside the fences:", sorted(outliers.tolist()))
```

```python
# Question 3

mean = days.mean()
median = days.median()
sd = days.std(ddof=1)
minimum = days.min()
maximum = days.max()

summary = pd.DataFrame({
    "Statistic": [
        "Mean",
        "Median",
        "Sample standard deviation",
        "First quartile",
        "Third quartile",
        "Minimum",
        "Maximum",
        "Interquartile range",
    ],
    "Days": [mean, median, sd, q1, q3, minimum, maximum, iqr],
})
summary["Days"] = summary["Days"].round(2)
summary.set_index("Statistic")
```

```python
# Question 4

hospital["observation"] = range(1, len(hospital) + 1)
hospital["z_score"] = (days - mean) / sd

pd.set_option("display.max_rows", 100)
hospital
```

```python
# Question 5

mu = 5
sigma = 3

p_less_than_10 = norm.cdf(10, mu, sigma)
p_between_3_and_10 = norm.cdf(10, mu, sigma) - norm.cdf(3, mu, sigma)

print("percent less than 10 days =", 100 * p_less_than_10)
print("percent between 3 and 10 days =", 100 * p_between_3_and_10)

se_mean = sigma / np.sqrt(35)
p_mean_over_6 = norm.sf(6, mu, se_mean)
print("P(sample mean > 6), n = 35 =", p_mean_over_6)
```

```python
# Question 6a

heights = np.array([
    156, 156, 157, 167, 168, 168, 169, 171, 171.5, 172,
    173, 174, 176, 177, 177, 178, 178, 178, 181, 182,
    183, 183, 184, 184, 185, 186, 187, 188, 189, 189, 189.5,
])

height_mean = heights.mean()
height_sd = heights.std(ddof=1)

print("mean =", height_mean)
print("sample standard deviation =", height_sd)

pd.DataFrame({
    "student": range(1, len(heights) + 1),
    "height_cm": heights,
    "z_score": (heights - height_mean) / height_sd,
})
```

```python
# Question 6b and 6c

p_one_student = norm.sf(190, height_mean, height_sd)
print("P(one student > 190 cm) =", p_one_student)

class_se = height_sd / np.sqrt(len(heights))
p_class_mean = norm.sf(190, height_mean, class_se)
print("P(class mean >= 190 cm) =", p_class_mean)
```
