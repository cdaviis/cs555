```python
# Setup

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t

%matplotlib inline
```

```python
# Question 1

calories = pd.read_csv("meal_calories.csv")
participants = calories.loc[calories["Group"] == "Participant", "Calories"]
nonparticipants = calories.loc[calories["Group"] == "Non-participant", "Calories"]

summary = pd.DataFrame({
    "n": [len(participants), len(nonparticipants)],
    "Mean": [participants.mean(), nonparticipants.mean()],
    "Sample SD": [participants.std(ddof=1), nonparticipants.std(ddof=1)],
    "Median": [participants.median(), nonparticipants.median()],
    "Min": [participants.min(), nonparticipants.min()],
    "Max": [participants.max(), nonparticipants.max()],
}, index=["Participant", "Non-participant"])
summary = summary.round(2)
print(summary.to_string())

bins = np.arange(100, 801, 100)
fig, axes = plt.subplots(1, 2, figsize=(8, 3.5), sharex=True, sharey=True)
axes[0].hist(participants, bins=bins, edgecolor="black")
axes[0].set_title("Participants")
axes[0].set_xlabel("Calories")
axes[0].set_ylabel("Number of children")
axes[1].hist(nonparticipants, bins=bins, edgecolor="black")
axes[1].set_title("Non-participants")
axes[1].set_xlabel("Calories")
fig.tight_layout()
fig.savefig("meal_calories_hist.png")
plt.show()
```

```python
# Question 2

from scipy.stats import t

calories = pd.read_csv("meal_calories.csv")
participants = calories.loc[calories["Group"] == "Participant", "Calories"]

n = len(participants)
mean = participants.mean()
sd = participants.std(ddof=1)
se = sd / np.sqrt(n)
t_stat = (mean - 425) / se
df = n - 1
p_value = 2 * t.sf(abs(t_stat), df)
t_crit = t.ppf(0.975, df)
t_90 = t.ppf(0.95, df)
lower = mean - t_90 * se
upper = mean + t_90 * se

print("n =", n)
print("mean =", f"{mean:.2f}")
print("sample sd =", f"{sd:.2f}")
print("standard error =", f"{se:.2f}")
print("t =", f"{t_stat:.3f}")
print("df =", df)
print("two-sided critical t =", f"{t_crit:.3f}")
print("two-sided p-value =", f"{p_value:.4f}")
print("90% CI =", f"{lower:.2f}", "to", f"{upper:.2f}")
```

```python
# Question 4

from scipy.stats import t

calories = pd.read_csv("meal_calories.csv")
participants = calories.loc[calories["Group"] == "Participant", "Calories"]
nonparticipants = calories.loc[calories["Group"] == "Non-participant", "Calories"]

n1 = len(participants)
n2 = len(nonparticipants)
mean1 = participants.mean()
mean2 = nonparticipants.mean()
s1 = participants.std(ddof=1)
s2 = nonparticipants.std(ddof=1)
v1 = s1**2 / n1
v2 = s2**2 / n2
se = np.sqrt(v1 + v2)
diff = mean1 - mean2
t_stat = diff / se
df = (v1 + v2)**2 / (v1**2 / (n1 - 1) + v2**2 / (n2 - 1))
p_value = t.sf(t_stat, df)
t_crit = t.ppf(0.95, df)
lower = diff - t_crit * se
upper = diff + t_crit * se

print("participant mean =", f"{mean1:.2f}", " n =", n1, " sd =", f"{s1:.2f}")
print("non-participant mean =", f"{mean2:.2f}", " n =", n2, " sd =", f"{s2:.2f}")
print("difference =", f"{diff:.2f}")
print("standard error =", f"{se:.2f}")
print("t =", f"{t_stat:.3f}")
print("df =", f"{df:.3f}")
print("one-sided critical t =", f"{t_crit:.3f}")
print("one-sided p-value =", f"{p_value:.4f}")
print("90% CI for the difference =", f"{lower:.2f}", "to", f"{upper:.2f}")
```
